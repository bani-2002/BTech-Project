```python
from statsmodels.tsa.arima.model import ARIMA
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load and prepare your data
df = pd.read_csv("weather_timeseries.csv", parse_dates=['timestamp'], index_col='timestamp')
temp_data = df['temperature']
```

```python
# Fit ARIMA model
model = ARIMA(temp_data, order=(5,1,0))  # adjust (p,d,q) as needed
model_fit = model.fit()
```

```python
# Forecast 7 days
forecast = model_fit.forecast(steps=7)
print(forecast)
```

```python
# Plot
temp_data.plot(label='Observed')
forecast.plot(label='Forecast', color='red')
plt.legend()
plt.title("7-Day Temperature Forecast")
plt.show()
```
